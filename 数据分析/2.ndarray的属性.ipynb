```python
import numpy as np
```

```python
arr = np.array(1)#0维数组
print(arr)
print('数组的形状:',arr.shape)
print('数组的维度:',arr.ndim)
print('数组的元素类型:',arr.dtype)
print('数组的元素个数:',arr.size)
print("数组的转置:",arr.T)
```

```python
arr = np.array([1,2.5,3])#1维数组
print(arr)
print('数组的形状:',arr.shape)
print('数组的维度:',arr.ndim)
print('数组的元素类型:',arr.dtype)
print('数组的元素个数:',arr.size)
print("数组的转置:", arr.T)
```
