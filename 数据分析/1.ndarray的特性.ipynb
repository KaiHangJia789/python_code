1.ndarray的特性
多维度

```python
import numpy as np
```

```python
arr = np.array(1)#0维数组
print(arr)
print('arr的维度:',arr.ndim)
```

```python
arr = np.array([1,2,3])#1维数组
print(arr)
print('arr的维度:',arr.ndim)
```

```python
arr = np.array([[1,2,3],[4,5,6]])#2维数组
print(arr)
print('arr的维度:',arr.ndim)
```

同质性

```python
arr = np.array([1,'hello'])#1维数组，元素类型为字符串
print(arr)
```

```python
arr = np.array([1,2.5])#1维数组，元素类型为浮点数
print(arr)
```
