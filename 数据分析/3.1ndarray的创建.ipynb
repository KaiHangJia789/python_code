```python
import numpy as np
```

```python
#基础的创建方法
arr = [1,2,3]
arr = np.array(arr,dtype=np.float32)
print(arr.ndim) #维度
print(arr)
```

```python
#copy方法创建
arr2 = arr.copy()#深拷贝
print(arr2)
```

```python
#预定义数组
arr = np.zeros((2,3)) #创建一个包含6个零的数组
print(arr)
```

```python
#全一
arr = np.ones((2,3),dtype=np.int32) #创建一个包含6个一的数组
print(arr)
```

```python
#未初始化
arr = np.empty((2,3)) #创建一个包含6个未初始化的数组
print(arr)
```

```python
arr = np.empty([3,5])#创建一个包含15个未初始化的数组
print(arr)
```

```python
#np.full
arr = np.full((2,3),5)#创建一个二维数组，元素全为5
print(arr)
```

```python
#np.zeros_like(arr)#返回arr的零矩阵
arr = np.zeros_like(arr)#创建一个和arr一样大小的零矩阵
print(arr)
```
