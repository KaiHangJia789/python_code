```python
import numpy as np
```

```python
#特殊矩阵
#单位矩阵:主对角线上元素为1，其他元素为0的矩阵
arr = np.eye(5,dtype=np.int32) #创建一个5行5列的单位矩阵
print(arr)
```

```python
arr = np.eye(3,4,dtype=np.int32) #创建一个3行4列的单位矩阵
print(arr)
```

```python
#对角矩阵:主对角线上元素为指定值，其他元素为0的矩阵
arr = np.diag([1,2,3,4]) #创建一个对角矩阵，参数为一个列表，列表中的元素为主对角线上的元素
print(arr)
```

```python
#生成随机数组
#生成0~1之间的随机浮点数(均匀分布)
arr = np.random.rand(2,3) #创建一个2行3列的随机数组
print(arr)
```

```python
#生成指定范围内的随机浮点数
arr = np.random.uniform(3,4,5) #创建一个包含5个元素的随机数组，元素的值在0~1之间
print(arr)
```

```python
##生成指定范围内的随机整数
arr = np.random.randint(0,10,5)
print(arr)
```

```python
#生成随机数列(正态分布) 
arr = np.random.randn(2,3)
print(arr)
```

```python
#设置随机种子
np.random.seed(20) #设置随机种子为20
arr = np.random.randint(1,10,size=(2,5))
print(arr)
np.random.seed() #取消随机种子设置
```
