```python
import numpy as np
```

```python
a = np.array([1, 2, 3])
b = np.array([4, 5, 6])
print(a+b)
print(a-b)
print(a*b)
print(a/b)
```

```python
c = [1, 2, 3]
d = [4, 5, 6]
print(c+d)
```

```python
arr = np.array([[1, 2, 3], [4, 5, 6]])
print(arr+3)
print(arr*3)
```

```python
#广播机制:从后往前看维度，要么相等，要么有一个是 1
a = np.array([1,2,3])#1*3
b = np.array([[4],[5],[6]])#3*1
print(a+b)
```

```python
a = np.array([[1, 2, 3],[4, 5, 6]])
b = np.array([[4,5,6],[7,8,9]])
c = np.array([[4, 5],[7, 8],[10, 11]])
#对应的数相乘:a的元素个数要等于b的元素个数
print(a*b)
#矩阵相乘:a的行数要等于b的列数
print(a @ c)
print(a.dot(c))
```
