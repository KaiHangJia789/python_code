```python
import numpy as np
```

```python
#等差数列 2 4 6 8 10
#参数：start,stop,step
arr = np.arange(2,11,2) #创建一个等差数列，
print(arr)
```

```python
#等间隔数列
arr = np.linspace(0,1,5) 
#创建一个等间隔数列，
#参数：start,stop,num:创建的元素个数
print(arr)
```

```python
arr = np.linspace( 0,1,5,dtype=np.int32)
print(arr)
```

```python
#对数数列
arr = np.logspace(0,2,5,base=10)
#参数：start,stop,num:创建的元素个数,base:底数
print(arr)
```
